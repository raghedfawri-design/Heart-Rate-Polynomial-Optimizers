```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
file_path = "/kaggle/input/datasets/arashnic/fitbit/mturkfitbit_export_3.12.16-4.11.16/Fitabase Data 3.12.16-4.11.16/heartrate_seconds_merged.csv"
df = pd.read_csv(file_path)

display(df.head())

df_sample = df.head(1000).copy()

X = np.arange(len(df_sample))
Y = df_sample['Value'].values

X = (X - X.mean()) / X.std()
Y = (Y - Y.mean()) / Y.std()
```

```python
degree = 3
X_poly = np.vstack([X**i for i in range(degree + 1)]).T

print("Shape of X_poly:", X_poly.shape)
```

```python
def gradient_descent(X, Y, learning_rate=0.01, epochs=1000):
    m, n = X.shape
    weights = np.zeros(n)
    loss_history = []
    
    for epoch in range(epochs):
        Y_pred = np.dot(X, weights)
        loss = (1 / (2 * m)) * np.sum((Y_pred - Y) ** 2)
        loss_history.append(loss)
        
        dw = (1 / m) * np.dot(X.T, (Y_pred - Y))
        weights -= learning_rate * dw
        
    return weights, loss_history

learning_rate = 0.05
epochs = 1500
optimal_weights, losses = gradient_descent(X_poly, Y, learning_rate, epochs)

print("Final Loss for GD:", losses[-1])
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(X, Y, color='gray', alpha=0.5, label='Actual Data')
plt.plot(X, np.dot(X_poly, optimal_weights), color='red', linewidth=2, label='GD Polynomial Fit')
plt.xlabel('Scaled Time')
plt.ylabel('Heart Rate')
plt.legend()
plt.title('Gradient Descent - Polynomial Fit')

plt.subplot(1, 2, 2)
plt.plot(losses, color='blue')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.title('GD Loss Reduction over Epochs')

plt.show()
```
