```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
file_path = "/kaggle/input/datasets/arashnic/fitbit/mturkfitbit_export_3.12.16-4.11.16/Fitabase Data 3.12.16-4.11.16/heartrate_seconds_merged.csv"
df = pd.read_csv(file_path)

display(df.head())

df_sample = df.head(1000).copy()

X = np.arange(len(df_sample))
Y = df_sample['Value'].values

X = (X - X.mean()) / X.std()
Y = (Y - Y.mean()) / Y.std()
```

```python
degree = 3
X_poly = np.vstack([X**i for i in range(degree + 1)]).T

print("Shape of X_poly:", X_poly.shape)
```

```python
def stochastic_gradient_descent(X, Y, learning_rate=0.01, epochs=50):
    m, n = X.shape
    weights = np.zeros(n)
    loss_history = []
    
    for epoch in range(epochs):
        # خلط البيانات عشوائياً بكل دورة في الـ SGD
        indices = np.arange(m)
        np.random.shuffle(indices)
        X_shuffled = X[indices]
        Y_shuffled = Y[indices]
        
        for i in range(m):
            xi = X_shuffled[i:i+1]
            yi = Y_shuffled[i:i+1]
            
            prediction = np.dot(xi, weights)
            error = prediction - yi
            
            # تحديث الأوزان بناءً على عينة واحدة فقط
            dw = np.dot(xi.T, error)
            weights -= learning_rate * dw.flatten()
            
        # حساب الخطأ العام بعد كل دورة كاملة
        Y_pred_all = np.dot(X, weights)
        epoch_loss = (1 / (2 * m)) * np.sum((Y_pred_all - Y) ** 2)
        loss_history.append(epoch_loss)
        
    return weights, loss_history

learning_rate = 0.01
epochs = 50
optimal_weights, losses = stochastic_gradient_descent(X_poly, Y, learning_rate, epochs)

print("Final Loss for SGD:", losses[-1])
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(X, Y, color='gray', alpha=0.5, label='Actual Data')
plt.plot(X, np.dot(X_poly, optimal_weights), color='green', linewidth=2, label='SGD Polynomial Fit')
plt.xlabel('Scaled Time')
plt.ylabel('Heart Rate')
plt.legend()
plt.title('Stochastic Gradient Descent - Polynomial Fit')

plt.subplot(1, 2, 2)
plt.plot(losses, color='orange')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.title('SGD Loss Reduction over Epochs')

plt.show()
```
