```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
file_path = "/kaggle/input/datasets/arashnic/fitbit/mturkfitbit_export_3.12.16-4.11.16/Fitabase Data 3.12.16-4.11.16/heartrate_seconds_merged.csv"
df = pd.read_csv(file_path)

display(df.head())

df_sample = df.head(1000).copy()

X = np.arange(len(df_sample))
Y = df_sample['Value'].values

X = (X - X.mean()) / X.std()
Y = (Y - Y.mean()) / Y.std()
```

```python
degree = 3
X_poly = np.vstack([X**i for i in range(degree + 1)]).T

print("Shape of X_poly:", X_poly.shape)
```

```python
def rmsprop_optimizer(X, Y, learning_rate=0.01, beta=0.9, epsilon=1e-8, epochs=100):
    m, n = X.shape
    weights = np.zeros(n)
    s = np.zeros(n)
    loss_history = []
    
    for epoch in range(epochs):
        prediction = np.dot(X, weights)
        error = prediction - Y
        dw = (1 / m) * np.dot(X.T, error)
        
        # RMSprop update logic
        s = beta * s + (1 - beta) * (dw ** 2)
        weights -= (learning_rate / (np.sqrt(s) + epsilon)) * dw
        
        epoch_loss = (1 / (2 * m)) * np.sum((prediction - Y) ** 2)
        loss_history.append(epoch_loss)
        
    return weights, loss_history

learning_rate = 0.02
epochs = 150
optimal_weights, losses = rmsprop_optimizer(X_poly, Y, learning_rate=learning_rate, epochs=epochs)

print("Final Loss for RMSprop:", losses[-1])
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(X, Y, color='gray', alpha=0.5, label='Actual Data')
plt.plot(X, np.dot(X_poly, optimal_weights), color='teal', linewidth=2, label='RMSprop Polynomial Fit')
plt.xlabel('Scaled Time')
plt.ylabel('Heart Rate')
plt.legend()
plt.title('RMSprop - Polynomial Fit')

plt.subplot(1, 2, 2)
plt.plot(losses, color='teal')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.title('RMSprop Loss Reduction over Epochs')

plt.show()
```
